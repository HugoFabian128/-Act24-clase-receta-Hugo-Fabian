#### Aplicación móvil bancaria

```python
# Simulación de IA para Aplicación Móvil Bancaria

# Inicialización de datos del usuario
nombre_usuario = "Hugo"
saldo = 1500.0  # Saldo inicial
pin_seguridad = "1234"
app_activa = True

print("--- BIENVENIDO A TU ASISTENTE BANCARIO INTELIGENTE ---")

# Autenticación interactiva usando while
intento_pin = ""
intentos_restantes = 3

while intento_pin != pin_seguridad and intentos_restantes > 0:
    print(f"Hola {nombre_usuario}, para continuar, necesito tu PIN de seguridad.")
    intento_pin = input("Introduce tu PIN (4 dígitos): ")

    if intento_pin != pin_seguridad:
        intentos_restantes -= 1 # Operador aritmético de resta
        if intentos_restantes > 0:
            print(f"PIN incorrecto. Te quedan {intentos_restantes} intentos.")
        else:
            print("Acceso bloqueado por seguridad. Contacta a soporte.")
            app_activa = False

# Menú principal de la IA (Ciclo infinito controlado)
while app_activa:
    print(f"\n[IA]: ¿En qué puedo ayudarte hoy, {nombre_usuario}?")
    print("1. Consultar Saldo | 2. Depositar | 3. Retirar | 4. Análisis de Gastos | 5. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        # Consulta de saldo simple
        print(f"[IA]: Tu saldo actual disponible es de ${saldo}.")
        if saldo < 200:
            print("[IA]: Advertencia: Tu saldo es bajo. ¡Cuida tus gastos!")

    elif opcion == "2":
        # Depósito con operador aritmético
        monto_deposito = float(input("[IA]: ¿Cuánto deseas depositar?: "))
        if monto_deposito > 0:
            saldo += monto_deposito # Suma acumulativa [4]
            print(f"[IA]: Depósito exitoso. Tu nuevo saldo es ${saldo}.")
        else:
            print("[IA]: El monto debe ser un número positivo.")

    elif opcion == "3":
        # Retiro con lógica de validación relacional [Artifact 1, 110]
        monto_retiro = float(input("[IA]: Ingresa el monto a retirar: "))

        # Operadores lógicos y relacionales para verificar fondos
        if monto_retiro > saldo:
            print(f"[IA]: Error. No tienes fondos suficientes. Tu saldo es ${saldo}.")
        elif monto_retiro <= 0:
            print("[IA]: Operación no válida. Introduce un monto real.")
        else:
            saldo -= monto_retiro # Resta aritmética [5, 6]
            print(f"[IA]: Retiro procesado. Saldo restante: ${saldo}.")

    elif opcion == "4":
        # Simulación de análisis inteligente basado en rangos
        print("[IA]: Analizando tu comportamiento financiero...")
        if saldo >= 5000:
            print("[IA]: Tienes un perfil de ahorrador excelente. ¡Sigue así!")
        elif 1000 <= saldo < 5000:
            print("[IA]: Tu salud financiera es estable.")
        else:
            print("[IA]: Recomiendo limitar gastos innecesarios este mes.")

    elif opcion == "5":
        # Condición de salida para romper el ciclo while [4]
        print(f"[IA]: Gracias por usar nuestra banca móvil. ¡Hasta pronto, {nombre_usuario}!")
        app_activa = False

    else:
        print("[IA]: Opción no reconocida. Por favor, elige un número del 1 al 5.")
```
